## Esercizio 2: Trova la Temperatura Giusta

Genera la stessa frase con 5 temperature diverse (da 0.0 a 1.5).
Osserva come cambia la creatività della risposta.

```python
domanda = "Inventa un nome per un robot."
prompt = f"<|system|>\nRispondi in italiano.</s>\n<|user|>\n{domanda}</s>\n<|assistant|>\n"

# TODO: aggiungi altri valori di temperatura nella lista
temperature_da_testare = [0.0, 0.5]  # ← aggiungi 1.0, 1.2, 1.5

for temp in temperature_da_testare:
    cfg = ov_genai.GenerationConfig()
    cfg.max_new_tokens = 20
    cfg.temperature = temp
    cfg.do_sample = temp > 0
    
    tokens = []
    pipe.generate(prompt, cfg, lambda t: (tokens.append(t), False)[1])
    risposta = ''.join(tokens).strip()
    print(f"temp={temp:.1f} → {risposta[:60]}")
```

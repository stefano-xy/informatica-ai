Scarica TinyLlama-1.1B-Chat in formato OpenVINO e lo salva in `/workspace/models/`.

---

```python
from pathlib import Path
from huggingface_hub import snapshot_download

MODEL_DIR = Path("/workspace/models/tinyllama-chat-ov")
HF_MODEL_ID = "OpenVINO/TinyLlama-1.1B-Chat-v1.0-int8-ov"

if MODEL_DIR.exists() and any(MODEL_DIR.iterdir()):
    print(f"✅ Modello già presente in: {MODEL_DIR}")
else:
    print(f"⬇️  Download di '{HF_MODEL_ID}' ...")
    snapshot_download(
        repo_id=HF_MODEL_ID,
        local_dir=str(MODEL_DIR),
        ignore_patterns=["*.msgpack", "*.h5", "flax_*"],
    )
    print(f"\n✅ Modello salvato in: {MODEL_DIR}")
```

```python
# Verifica: elenca i file del modello
print("File nel modello:")
for f in sorted(MODEL_DIR.iterdir()):
    size_mb = f.stat().st_size / 1024
    print(f"  {f.name:40s}  {size_mb:7.1f} KB")
```
